```python
from data.dataset import ReimburseGraphDataset, StandardGraphDataset, DataAugmentationLevel, NodeType, DialogNode, Question
```

```python
import json
```

```python
ds = StandardGraphDataset('de/reimburse/traintest_graph.json', "de/reimburse/traintest_graph.json", False, augmentation=DataAugmentationLevel.NONE, augmentation_path=None, resource_dir="resources")
answer_filename = "resources/de/reimburse/traintest_answers.json"
with open(answer_filename, "r") as f:
    synonyms = json.load(f)
```

```python
new_answers = {}

for node in ds.nodes_by_type[NodeType.QUESTION]:
    for answer in node.answers:
        ans_synonyms = set([answer.text])
        ans_synonyms = ans_synonyms.union(synonyms[answer.text.lower()])
        new_answers[answer.key] = list(ans_synonyms)

print(len(new_answers))
```

```python
with open(answer_filename, "w") as f:
    json.dump(new_answers, f)
```

```python
ds = ReimburseGraphDataset('de/reimburse/traintest_graph.json', "de/reimburse/traintest_answers.json", True, augmentation=DataAugmentationLevel.NONE, augmentation_path=None, resource_dir="resources")
```

```python
import time

generated = {}
with open("resources/en/reimburse/generated/chatgpt/train_questions_v1.json", "r") as f:
    data = json.load(f)
    for node_key in data:
        node = ds.nodes_by_key[int(node_key)]
        for question in data[node_key]:
            key = str(time.time()).replace(".", "")
            generated[key] = {
                "dialog_node_key": node_key,
                "text": question,
                "node_text": node.text,
                "node_type": node.node_type.value
            }
```

```python
with open("resources/en/reimburse/generated/chatgpt/train_questions_v1.json", "w") as f:
    json.dump(generated, f)
```
