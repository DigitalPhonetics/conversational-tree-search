```python
import json
```

```python
with open('resources/en/augmentation/augmentation_with_judgement.json', 'r') as f:
    data = json.load(f)

    negative_judgement = set()
    positive_judgement = set()

    for candidate in data:
        if not data[candidate]['judgement'].lower().startswith("yes.") and not data[candidate]['judgement'].lower().startswith('yes,') and not ("the answer is yes") in data[candidate]['judgement'].lower() and not "the given text answers the question" in data[candidate]['judgement'].lower() and not "the answer is:" in data[candidate]['judgement'].lower():
            negative_judgement.add(candidate)
        else:
            positive_judgement.add(candidate)

    print(f"{len(negative_judgement)} negative samples")
    print(f"{len(positive_judgement)} positive samples")
    
for candidate in negative_judgement:
    del data[candidate]
print(f"{len(data)} samples to write")

with open('resources/en/augmentation/augmentation_info_nodes_filtered.json', 'w') as f:
    json.dump(data, f)
```
